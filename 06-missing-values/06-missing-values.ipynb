# 06 - Handling Missing Values


Real data contains **missing** (`NaN`) and **infinite** (`Inf`) values. NumPy gives you tools to detect and fix them.

| Function | Returns `True` when the value is... |
|---|---|
| `np.isnan(x)` | Not a Number (`NaN`) |
| `np.isinf(x)` | positive or negative infinity |
| `np.isfinite(x)` | **finite** (not NaN and not Inf) |
| `np.nan_to_num(x)` | replaces NaN → 0, Inf → large finite numbers |

### Key gotcha

`NaN == NaN` is `False` — you **cannot** compare with `==` to find missing values. You must use `np.isnan` (or a mask).

### Strategies

1. **Detect** — `np.isnan`, `np.isinf`.
2. **Remove** — keep only finite values with `arr[np.isfinite(arr)]`.
3. **Replace** — `np.nan_to_num`, or fill with a statistic (mean, median...).
4. **Aggregate safely** — `np.nanmean`, `np.nansum`, `np.nanstd` skip the missing values instead of returning `NaN`.

```python
import numpy as np

arr = np.array([1, 2, np.nan, 4, 5, np.nan, 7])
arr
```

## 1. Detecting missing values

`np.isnan` builds a boolean mask of where the missing values are.

```python
np.isnan(arr)
```

```python
# Why not ==? Because NaN never equals itself
print("NaN == NaN is", np.nan == np.nan)
print("np.isnan(np.nan) is", np.isnan(np.nan))
```

## 2. Replacing with `nan_to_num`

`nan_to_num` turns `NaN` into 0 (and infinities into large finite numbers).

```python
np.nan_to_num(arr)
```

## 3. Filtering out missing values

Use `~` (NOT) on the mask to keep only the valid values.

```python
clean = arr[~np.isnan(arr)]
clean
```

## 4. Infinity

`isinf` and `isfinite` deal with `Inf` values.

```python
with_inf = np.array([1, 2, np.inf, 4, -np.inf, 6])
np.isinf(with_inf)
```

```python
np.isfinite(with_inf)   # True only for normal numbers
```

```python
# Remove both NaN and Inf in one go
dirty = np.array([1.0, np.nan, 2.0, np.inf, 3.0])
dirty[np.isfinite(dirty)]
```

## 5. Safe aggregations

Ordinary `mean` returns `NaN` when data contains missing values; the `nan`-variants skip them.

```python
print("mean  (ignores nothing):", np.mean(arr))
print("nanmean (skips NaN)    :", np.nanmean(arr).round(3))
print("nansum (skips NaN)     :", np.nansum(arr))
print("nanstd (skips NaN)     :", np.nanstd(arr).round(3))
```

## 6. Filling with a statistic

Replace missing values with the mean of the valid ones.

```python
filled = np.where(np.isnan(arr), np.nanmean(arr), arr)
filled
```

```python
# Verify: no NaN left
np.isnan(filled).sum()
```
