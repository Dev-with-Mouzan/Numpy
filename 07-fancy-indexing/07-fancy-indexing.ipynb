# 07 - Fancy Indexing

**Fancy indexing** means selecting elements with arrays of *indices* instead of slices. It can be **much more flexible** than slicing:

| Technique | Example | Returns |
|---|---|---|
| Integer array index | `a[[0, 2, 5]]` | new array of those positions |
| `np.take(a, idx)` | same as above, different spelling | new array |
| `np.put(a, idx, vals)` | write into chosen positions | modifies `a` in place |
| 2D index arrays | `a[[0,1],[1,2]]` | pairs `(0,1)`, `(1,2)` |
| `np.select([c1,c2],[x,y])` | many conditions at once | element-wise choice |
| `np.clip(a, lo, hi)` | bound values | trimmed values |
| `np.searchsorted(a, v)` | where `v` would slot into sorted `a` | integer positions |

### Key difference from slicing

Fancy indexing **always copies** the data (unlike slices, which are views). So `b = a[[0,1]]` is safe to modify.

### Boolean + integer combined

Booleans pick *which rows*, integers pick *which columns* — e.g. `data[mask][:, [0, 3]]`.

```python
import numpy as np

a = np.arange(10)
a
```

## 1. Integer array indexing

Select arbitrary positions with a list of indices — the result is a **new array**.

```python
a[[0, 2, 5, 9]]
```

```python
np.take(a, [3, 3, 3, 1])   # repetition allowed
```

## 2. Writing with `np.put`

`np.put` modifies the original array in place at the chosen positions.

```python
b = np.zeros(6, dtype=int)
np.put(b, [0, 2, 5], [7, 8, 9])
b
```

## 3. 2D fancy indexing

Two equal-length index arrays pair up to pick `(row, col)` coordinates.

```python
m = np.arange(12).reshape(3, 4)
m
```

```python
m[[0, 1, 2], [1, 3, 0]]   # (0,1), (1,3), (2,0) are index of the element
```

```python
# A list of row indices reorders the rows
m[[2, 0, 1]] # first provided the rows then colums
```

## 4. Fancy indexing copies (unlike slices)

Changing a fancy-indexed result never touches the source array.

```python
sub = m[[0, 1]]
sub[0, 0] = -99
print("sub changed:")
print(sub)
print("original untouched:")
print(m)
```

## 5. `np.select` — many conditions at once

Give a list of conditions and a matching list of results; `default` fills the rest.

```python
scores = np.array([10, 45, 80, 95, 62])
np.select(
    [scores >= 90, scores >= 70, scores >= 50],
    ["A", "B", "C"],
    default="F",
)
```

## 6. `np.clip` — bound the values

Clamp every value to a range — handy for removing outliers before plotting.

```python
data = np.array([-5, 2, 7, 99, 3, -20])
np.clip(data, 0, 10)
```

## 7. `np.searchsorted` — find insertion points

For a **sorted** array, tells you where values would slot in to keep the order.

```python
sorted_a = np.array([1, 3, 5, 7, 9])
np.searchsorted(sorted_a, [4, 8, 0])
```

```python
# Use the positions to build a binned histogram quickly
np.searchsorted([0, 50, 100], [10, 70, 120, -3, 50])
```

## 8. Combining booleans and integers

Keep matching rows, then pick specific columns.

```python
table = np.arange(20).reshape(5, 4)
table
```

```python
table[table[:, 0] % 2 == 0][:, [0, 2]]   # even first-column rows, cols 0 & 2
```

## Summary

- Index with integer arrays (`a[[0,2,5]]`, `np.take`, 2D pairs).
- Write with `np.put`; select many conditions with `np.select`; bound with `np.clip`.
- `np.searchsorted` locates positions in sorted arrays.
- Fancy indexing **copies** (unlike slicing views).
