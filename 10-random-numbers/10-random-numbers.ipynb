# 10 - Random Numbers & Simulation

NumPy's modern random API is **`np.random.default_rng()`** — a generator object you create once and reuse.

### Why a generator object?

```python
rng = np.random.default_rng(42)   # 42 = seed -> reproducible
rng.normal(size=(3, 3))           # same numbers every run
```

A **seed** makes randomness reproducible — essential for tests and teaching.

### Common distributions / methods

| Method | Produces |
|---|---|
| `rng.normal(mu, sigma, size)` | bell-shaped continuous values |
| `rng.uniform(lo, hi, size)` | flat random values |
| `rng.integers(lo, hi, size)` | random integers (hi exclusive) |
| `rng.binomial(n, p, size)` | number of successes |
| `rng.choice(a, size, p=...)` | pick elements, optionally with weights |
| `rng.permutation(a)` | shuffled copy |
| `rng.shuffle(a)` | shuffle in place |

### Simulation ideas

- **Random walk**: cumsum of random ±1 steps.
- **Monte Carlo π**: random points in a square, fraction inside a circle.

````python
import numpy as np

rng = np.random.default_rng(42)   # seeded generator
````

## 1. Reproducibility

Two generators with the same seed produce identical sequences.

````python
g1 = np.random.default_rng(1)
g2 = np.random.default_rng(1)
print(g1.integers(0, 100, 5))
print(g2.integers(0, 100, 5))
````

## 2. Distributions

Generate from common distributions and inspect their stats.

````python
normal = rng.normal(loc=100, scale=10, size=10_000)
print("normal: mean", normal.mean().round(2), "std", normal.std().round(2))
````

````python
uniform = rng.uniform(0, 1, size=10_000)
print("uniform: min", uniform.min().round(3), "max", uniform.max().round(3))
````

````python
rolls = rng.integers(1, 7, size=12)   # dice rolls (1..6)
rolls
````

````python
flips = rng.binomial(n=10, p=0.5, size=20)   # heads in 10 coin flips, 20 trials
flips
````

## 3. Sampling and shuffling

`choice` picks elements (with or without weights); `permutation`/`shuffle` reorder.

````python
items = np.array(["apple", "banana", "cherry"])
rng.choice(items, size=6, p=[0.6, 0.3, 0.1])   # weighted sampling
````

````python
deck = np.arange(52)
shuffled = rng.permutation(deck)   # shuffled copy
shuffled[:8]
````

````python
rng.shuffle(deck)   # shuffles in place
deck[:8]
````

## 4. Random walk

Cumulative sum of random ±1 steps models a drunkard's walk / stock moves.

````python
steps = rng.choice([-1, 1], size=100)
walk = np.cumsum(steps)
print("final position:", walk[-1])
print("max excursion  :", walk.max(), walk.min())
walk[:15]
````

## 5. Monte Carlo estimation of pi

Throw random points into a unit square; the fraction landing in the quarter-circle approximates `pi / 4`.

````python
n = 100_000
x = rng.uniform(0, 1, n)
y = rng.uniform(0, 1, n)
inside = (x**2 + y**2) <= 1
pi_est = 4 * inside.sum() / n
print(f"estimated pi: {pi_est:.4f}   (true pi: {np.pi:.4f})")
````

## 6. Bootstrapping a statistic

Resample with replacement many times to estimate the uncertainty of a mean.

````python
sample = rng.normal(50, 5, size=200)
boot = np.array([rng.choice(sample, size=200, replace=True).mean() for _ in range(1000)])
print("sample mean      :", sample.mean().round(2))
print("bootstrap mean   :", boot.mean().round(2))
print("95% CI           :", np.percentile(boot, [2.5, 97.5]).round(2))
````

## Summary

- Use `np.random.default_rng(seed)` for **reproducible** randomness.
- `normal`, `uniform`, `integers`, `binomial`, `choice`, `permutation`, `shuffle`.
- Power tools: random walks (`cumsum`), Monte Carlo, bootstrapping.
