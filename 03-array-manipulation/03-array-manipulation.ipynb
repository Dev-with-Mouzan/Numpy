# 03 - Array Manipulation


Reshaping and combining arrays is how you prepare data for a computation.

| Operation | Function / method | Result |
|---|---|---|
| Reshape | `a.reshape((r, c))` | same data, new shape |
| Flatten | `a.ravel()` / `a.flatten()` | 1D array |
| Transpose | `a.T` | rows <-> columns |
| Concatenate | `np.concatenate([a, b])` | join arrays |
| Stack rows | `np.vstack([a, b])` | one under the other |
| Stack columns | `np.hstack([a, b])` | side by side |
| Reverse | `np.flip(a)` | elements in reverse order |
| Split | `np.array_split(a, n)` | divide into parts |

### `ravel` vs `flatten`

- `ravel()` returns a **view** when possible (fast, shares memory).
- `flatten()` always returns a **copy**.

### Reshape trick
You can let NumPy compute a dimension for you with `-1`: `a.reshape(2, -1)` fits the rest automatically (total size must match).

```python
import numpy as np

array = np.arange(10)
array
```

## 1. Reshaping

`reshape((2, 5))` reinterprets the same 10 values as 2 rows of 5.

```python
new_array = array.reshape((2, 5))
new_array
```

```python
# Let NumPy work out one dimension with -1
array.reshape(5, -1)
```

## 2. Back to 1D

`ravel()` and `flatten()` both turn the array back into one dimension.

```python
new_array.ravel()   # or new_array.flatten()
```

```python
new_array.flatten()
```

## 3. Transposing

`a.T` swaps rows and columns.

```python
mat = np.array([[1, 2, 3], [4, 5, 6]])
print("original shape:", mat.shape)
print("transposed shape:", mat.T.shape)
mat.T
```

## 4. Combining arrays

`concatenate` joins 1D arrays end-to-end.

```python
a1 = np.array([1, 2, 3])
a2 = np.array([4, 5, 6])
np.concatenate((a1, a2))
```

## 5. Stacking 2D arrays

`vstack` stacks **vertically** (rows); `hstack` stacks **horizontally** (columns).

```python
m1 = np.array([[1, 2, 3], [4, 5, 6]])
m2 = np.array([[7, 8, 9], [10, 11, 12]])
```

```python
np.vstack((m1, m2))   # one under the other
```

```python
np.hstack((m1, m2))   # side by side
```

## 6. Reversing and splitting

`np.flip` reverses an array (it is a function, not a method).

```python
np.flip(a1)          # reversed
```

```python
np.array_split(a1, 3)  # split into 3 parts
```

## 7. Related attributes

A quick recap of the array's own numbers.

```python
print("ndim   :", mat.ndim)      # number of dimensions
print("size   :", mat.size)      # total elements
print("nbytes :", mat.nbytes)    # memory used
```

## Summary

- `reshape((r, c))` (with `-1` magic), `ravel()`/`flatten()` for 1D, `a.T` to transpose.
- Combine with `concatenate`, `vstack`, `hstack`; reverse with `np.flip`.
- Note: `flip` and `flatten` are **functions/methods with specific names** — `np.flip` is the function form.
