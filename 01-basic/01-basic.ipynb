# 01 - NumPy Basics

## What is NumPy?

**NumPy** (Numerical Python) is the foundation of scientific computing in Python. Its core object is the **`ndarray`** — a fast, fixed-size, multidimensional array.

### Why arrays instead of lists?

| | Python list | NumPy `ndarray` |
|---|---|---|
| Stores | mixed types allowed | one **dtype** for all elements |
| Element access | loop needed | **vectorised** (C speed) |
| Memory | scattered objects | **contiguous** block |
| Maths on whole list | error / slow loop | `a * 2` just works |

### Creating arrays

| Function | Creates |
|---|---|
| `np.array(list)` | array from Python data |
| `np.zeros(shape)` | all zeros |
| `np.ones(shape)` | all ones |
| `np.full(shape, v)` | all equal to `v` |
| `np.arange(n)` | 0, 1, 2, ... n-1 |
| `np.linspace(a, b, n)` | `n` evenly spaced values between `a` and `b` |
| `np.eye(n)` | `n x n` identity matrix |
| `np.random.default_rng()` | arrays of random numbers |

### Key attributes

`shape` (dimensions), `ndim` (number of dims), `dtype` (element type), `size` (total elements), `nbytes` (memory).

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 1. From list to array

A **1D array** from a Python list, then a **2D array** from a list of lists (rows).

```python
l = [1, 2, 3, 4, 5]
a = np.array(l)
a
```

```python
# Two dimensional array: a list of rows
b = np.array([[1, 2, 3], [4, 5, 6]])
b
```

## 2. Pre-filled arrays

Fast ways to build arrays without typing data.

```python
zero = np.zeros((3, 4))   # 3 rows, 4 columns, all zeros
zero
```

```python
one = np.ones((2, 3))     # 2 rows, 3 columns, all ones
one
```

```python
filled = np.full((2, 3), 10)  # all values equal to 10
filled
```

```python
empty = np.empty((2, 3))   # uninitialised (fast, but values are garbage)
empty
```

## 3. Ranges of values

`arange` gives consecutive integers; `linspace` gives evenly spaced values over a range.

```python
arr = np.arange(10)        # 0 to 9
arr
```

```python
lin = np.linspace(0, 5, 10)  # 10 values evenly spaced from 0 to 5
lin
```

## 4. Identity and random arrays

`eye` / `identity` build identity matrices; `default_rng` is the modern way to get random values.

```python
eye = np.eye(4)            # 4x4 identity matrix
eye
```

```python
ident = np.identity(5)     # 5x5 identity matrix
ident
```

```python
rng = np.random.default_rng(42)          # reproducible generator
rnd = rng.normal(size=(3, 3))            # 3x3 random normal values
rnd
```

## 5. Array attributes

Ask the array about itself: shape, dimension count, dtype, size and memory.

```python
custom = np.full((2, 3), 10)
print("shape :", custom.shape)
print("ndim  :", custom.ndim)
print("dtype :", custom.dtype)
print("size  :", custom.size)
print("nbytes:", custom.nbytes)
```

```python
# Change the element type
ints = np.array([1, 2, 3])
floats = ints.astype("float64")
print(ints.dtype, "->", floats.dtype)
```

## Summary

- NumPy arrays (`ndarray`) are **fast, vectorised, single-dtype** containers.
- Create them with `np.array`, `zeros`, `ones`, `full`, `arange`, `linspace`, `eye`, `default_rng`.
- Inspect them with `shape`, `ndim`, `dtype`, `size`, `nbytes`; change type with `.astype()`.
