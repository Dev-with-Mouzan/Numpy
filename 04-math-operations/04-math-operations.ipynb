# 04 - Mathematical Operations

NumPy makes maths on arrays **vectorised** — one expression replaces a whole Python loop.

### Element-wise arithmetic

`+`, `-`, `*`, `/`, `//`, `**` apply to **every element** at once (see 05-broadcasting for how shapes align).

### Universal functions (ufuncs)

`np.sqrt`, `np.exp`, `np.log`, `np.abs`, `np.sin`, ... run the same math on each element.

### Aggregations

| Function | Meaning |
|---|---|
| `np.sum` / `np.mean` | total / average |
| `np.min` / `np.max` | smallest / largest |
| `np.std` / `np.median` | spread / middle value |
| `np.argmin` / `np.argmax` | **index** of smallest / largest |
| `np.argsort` | index order for sorting |
| `np.count_nonzero` | how many non-zero values |

On 2D arrays add `axis=0` (down columns) or `axis=1` (across rows).

### Conditional selection

- `arr[condition]` → keep the matching **values** (boolean mask).
- `np.where(condition, a, b)` → element-wise `a` where true, `b` where false.

```python
import numpy as np

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
```

## 1. Arithmetic

Each operator works element by element.

```python
array1 + array2   # addition
```

```python
array2 - array1   # subtraction
```

```python
array1 * array2   # multiplication
```

```python
array2 / array1   # division
```

```python
array2 // array1  # floor division (integer quotient)
```

```python
array1 ** 2       # power
```

## 2. Universal functions

Math functions applied to every element.

```python
np.sqrt(array1)
```

```python
np.abs(np.array([-1, 2, -3]))
```

```python
np.exp(array1).round(3)
```

## 3. Aggregations

Reduce the whole array to a single number.

```python
print("sum   :", np.sum(array1))
print("mean  :", np.mean(array1))
print("min   :", np.min(array1))
print("max   :", np.max(array1))
print("std   :", np.std(array1))
print("median:", np.median(array1))
```

## 4. Aggregations with `axis`

On a 2D array, `axis=0` goes down columns, `axis=1` goes across rows.

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
print("column sums (axis=0):", m.sum(axis=0))
print("row sums    (axis=1):", m.sum(axis=1))
```

## 5. Positions of extremes

`argmin` / `argmax` return the **index** of the smallest / largest value.

```python
arr = np.array([4, 1, 9, 2])
print("array        :", arr)
print("argmin (idx) :", arr.argmin())
print("argmax (idx) :", arr.argmax())
```

## 6. Sorting

`np.sort` returns sorted values; `argsort` returns the index order that would sort the array.

```python
np.sort(arr)
```

```python
np.argsort(arr)   # indexes that put the array in order
```

## 7. Conditional selection

Two ways to pick values based on a condition.

```python
# Boolean mask: keep only values greater than 2
array1[array1 > 2]
```

```python
# np.where: choose values element by element (keep value, or 0)
np.where(array1 > 2, array1, 0)
```

```python
np.count_nonzero(array1)   # how many non-zero values
```

## Summary

- Arithmetic and ufuncs (`sqrt`, `abs`, `exp`) are **element-wise and vectorised**.
- Aggregations: `sum`, `mean`, `min`, `max`, `std`, `median` (+ `axis=` on 2D).
- `argmin`/`argmax`/`argsort` give positions; masks and `np.where` select by condition.
