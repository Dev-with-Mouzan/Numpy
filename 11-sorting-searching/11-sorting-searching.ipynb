# 11 - Sorting & Searching

NumPy provides fast, vectorised order and lookup operations.

### Sorting

| Function | Returns |
|---|---|
| `np.sort(a)` | sorted **copy** |
| `a.sort()` | sorts **in place** (method) |
| `np.argsort(a)` | index order that sorts `a` |
| `np.partition(a, k)` | elements around the `k`-th position (partial sort, faster) |
| `np.argpartition(a, k)` | indices for `np.partition` |

On 2D arrays use `axis=0` / `axis=1`.

### Searching

| Function | Finds |
|---|---|
| `np.searchsorted(a, v)` | where `v` fits into a sorted `a` |
| `np.argmin` / `np.argmax` | index of min / max |
| `np.where(cond)` | indices where `cond` is true |
| `np.nonzero(a)` | indices of non-zero elements |

### Unique & counting

- `np.unique(a, return_counts=True)` — distinct values + counts.
- `np.bincount(a)` — fast counts for non-negative integers (acts like a histogram).
- `np.in1d(a, b)` — which elements of `a` are in `b`.

```python
import numpy as np

a = np.array([3, 1, 4, 1, 5, 9, 2, 6])
a
```

## 1. Sorting

`np.sort` returns a sorted copy; the method `a.sort()` sorts in place.

```python
np.sort(a)
```

```python
np.sort(a)[::-1]   # descending
```

```python
print("original untouched:", a)
```

## 2. `argsort` — sort by another array

`argsort` gives the index order, so you can sort **related arrays together**.

```python
names = np.array(["c", "a", "b"])
values = np.array([30, 10, 20])
order = np.argsort(values)
print("order :", order)
print("names :", names[order])
print("values:", values[order])
```

## 3. Sorting 2D arrays

Choose the axis: rows or columns.

```python
m = np.array([[5, 2], [1, 9], [7, 3]])
np.sort(m, axis=0)   # sort down each column
```

```python
np.sort(m, axis=1)   # sort across each row
```

## 4. Partitioning — faster than full sort

`np.partition` guarantees the `k`-th element is in its sorted position; elements before it are all smaller, after it all larger — but not ordered internally.

```python
# The 3rd smallest element is at index 3; smaller ones before, larger after
np.partition(a, 3)
```

```python
# 3 largest values via argpartition
idx = np.argpartition(a, -3)[-3:]
np.sort(a[idx])
```

## 5. Searching

`where` / `nonzero` give the positions of matching elements.

```python
np.where(a > 4)
```

```python
np.nonzero(np.array([0, 5, 0, 7]))
```

```python
# Insertion positions in sorted data
np.searchsorted(np.sort(a), [4, 10])
```

## 6. Unique values and counting

`unique` with counts is a one-line histogram; `bincount` is the fast integer counter.

```python
vals, counts = np.unique(a, return_counts=True)
print("values:", vals)
print("counts:", counts)
```

```python
np.bincount(np.array([0, 3, 1, 3, 0, 3]))
```

```python
np.in1d(a, [2, 9])   # which elements of a are in [2, 9]
```

## Summary

- `np.sort` / `a.sort()` / `argsort` (sort by another array) / `partition` (faster partial sort).
- `where`, `nonzero`, `searchsorted`, `argmin`/`argmax` for searching.
- `np.unique(..., return_counts=True)` and `np.bincount` for counting.
