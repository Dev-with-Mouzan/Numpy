# 12 - Structured Arrays


A normal array holds many values of **one** type. A **structured array** holds **records** — rows with *named fields*, like a mini spreadsheet or a table in a single array.

```python
dtype = np.dtype([("name", "U10"), ("age", int), ("score", float)])
recs = np.array([("Ada", 36, 99.5)], dtype=dtype)
recs["age"]      # one column
recs[0]           # one record
```

### Common field types

| Code | Meaning |
|---|---|
| `i4` / `int64` | integers |
| `f8` / `float64` | floats |
| `b1` | booleans |
| `U10` | unicode string up to 10 chars |

### Structured vs record arrays

- A **structured array** uses `arr["field"]` (square brackets for columns).
- A **record array** (`np.rec.array` / `.view(np.recarray)`) also allows `arr.field` (dot access).

> Use structured arrays when you want one contiguous table with mixed types. Pandas DataFrames are the higher-level alternative.

````python
import numpy as np
````

## 1. Defining a dtype with fields

Describe the columns once; then build the array.

````python
person_dtype = np.dtype([
    ("name", "U10"),
    ("age", "i4"),
    ("score", "f8"),
])
person_dtype
````

````python
people = np.array([
    ("Ada", 36, 99.5),
    ("Bob", 41, 85.0),
    ("Cid", 22, 91.5),
], dtype=person_dtype)
people
````

## 2. Accessing fields and records

Field name = column; integer index = record (row).

````python
people["age"]
````

````python
people[1]            # one record
````

````python
people[people["score"] > 90]["name"]   # filter then pick a column
````

## 3. Operations on fields

Every field is just an array — all vectorised ops work on it.

````python
print("mean age  :", people["age"].mean())
print("max score :", people["score"].max())
print("oldest    :", people[people["age"].argmax()]["name"])
````

## 4. Sorting a structured array

Sort by a field with `np.sort(order=...)` (a named argument, not a field).

````python
np.sort(people, order="score")
````

## 5. Record arrays — dot access

`.view(np.recarray)` lets you use `people.name` instead of `people["name"]`.

````python
recs = people.view(np.recarray)
recs.name
````

````python
recs.score.mean()
````

## 6. Building a structured array from columns

Combine separate arrays into one table with `np.rec.fromarrays`.

````python
ids = np.array([1, 2, 3])
values = np.array([9.9, 8.8, 7.7])
table = np.rec.fromarrays([ids, values], names="id,value")
table
````

## Summary

- Structured arrays hold **named, mixed-type columns** in one array via a field dtype.
- Access columns with `arr["field"]`, rows with `arr[i]`; filter with masks.
- `.view(np.recarray)` adds dot access; `np.rec.fromarrays` builds tables from columns.
