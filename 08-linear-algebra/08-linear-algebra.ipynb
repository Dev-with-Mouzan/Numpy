# 08 - Linear Algebra


NumPy's `np.linalg` module brings fast **linear algebra** to Python.

### Matrix multiplication

| Form | What |
|---|---|
| `a * b` | element-wise (see 05-broadcasting) |
| `a @ b` / `np.matmul(a, b)` | **true matrix product** |
| `np.dot(a, b)` | dot product (matmul for 2D) |
| `np.inner` / `np.outer` | vector inner / outer product |

For matrices `(n,k) @ (k,m) -> (n,m)`. The `@` operator is the modern, readable way.

### Common `np.linalg` functions

| Function | Meaning |
|---|---|
| `det` | determinant (0 -> singular) |
| `inv` | inverse matrix |
| `solve(A, b)` | solve `A x = b` (faster & more stable than `inv @ b`) |
| `eig` | eigenvalues & eigenvectors |
| `norm` | vector / matrix magnitude |
| `trace` | sum of diagonal elements |
| `matrix_power` | integer powers of a matrix |

> **Rule:** prefer `np.linalg.solve` over `np.linalg.inv` when solving systems — it is numerically stable and faster.

```python
import numpy as np

A = np.array([[2.0, 1.0],
              [1.0, 3.0]])
B = np.array([[1.0, 2.0],
              [3.0, 4.0]])
v = np.array([1.0, 2.0])
```

## 1. Element-wise vs matrix multiplication

`*` multiplies element by element; `@` does the real matrix product.

```python
A * B     # element-wise
```

```python
A @ B     # matrix product
```

```python
A @ v     # matrix-vector product
```

## 2. Dot, inner, outer

`dot` is the classic product; `outer` builds a matrix from two vectors.

```python
np.dot(v, v)        # scalar dot product
```

```python
np.outer(v, v)      # 2x2 outer product matrix
```

## 3. Determinant and inverse

`det` tells whether a matrix is invertible; `inv` returns the inverse.

```python
np.linalg.det(A)
```

```python
A_inv = np.linalg.inv(A)
print("inverse:")
A_inv
print("A @ A_inv (should be identity):")
np.round(A @ A_inv, 2)
```

## 4. Solving linear systems

Solve `A x = b` for `x` with `np.linalg.solve`.

```python
b = np.array([5.0, 7.0])
x = np.linalg.solve(A, b)
print("solution x:", x)
print("check A @ x:", np.round(A @ x, 2))   # should equal b
```

## 5. Eigenvalues and eigenvectors

`eig` returns `(eigenvalues, eigenvectors)`. Each column of the vectors matrix is an eigenvector.

```python
w, vec = np.linalg.eig(A)
print("eigenvalues :", np.round(w, 3))
print("eigenvectors:")
np.round(vec, 3)
```

## 6. Norms and trace

`norm` measures size; `trace` sums the diagonal.

```python
print("L2 norm of v      :", np.linalg.norm(v).round(4))
print("L1 norm of v      :", np.linalg.norm(v, ord=1))
print("Frobenius norm of A:", np.linalg.norm(A).round(4))
print("trace of A        :", np.trace(A))
```

## 7. Matrix powers

`matrix_power` raises a square matrix to an integer power (e.g. Markov chains).

```python
np.linalg.matrix_power(B, 2)   # B @ B
```

## Summary

- `*` = element-wise; `@` / `np.matmul` = matrix product; `np.dot`, `np.inner`, `np.outer` for vector products.
- `np.linalg`: `det`, `inv`, `solve`, `eig`, `norm`, `matrix_power`.
- Prefer `solve(A, b)` over `inv(A) @ b`.
